```python
# Статистика для столбца popularity
popularity_stats = df['popularity'].describe()
mean_pop = df['popularity'].mean()
median_pop = df['popularity'].median()
min_pop = df['popularity'].min()
max_pop = df['popularity'].max()
std_pop = df['popularity'].std()
print(f"Среднее арифметическое: {mean_pop:.2f}")
print(f"Медиана: {median_pop:.2f}")
print(f"Минимум: {min_pop}")
print(f"Максимум: {max_pop}")
print(f"Стандартное отклонение: {std_pop:.2f}")
print(f"\nПолная описательная статистика:\n{popularity_stats}")
```

```python
# Построение гистограммы распределения popularity
plt.figure(figsize=(12, 7))

# Строим гистограмму
n, bins, patches = plt.hist(df['popularity'], bins=20, edgecolor='black', alpha=0.7, color='skyblue')

# Настройка графика
plt.title('Распределение популярности треков', fontsize=16, fontweight='bold')
plt.xlabel('Уровень популярности (popularity)', fontsize=14)
plt.ylabel('Количество треков', fontsize=14)
plt.grid(axis='y', alpha=0.3)

# Добавляем значения над столбцами
for i, count in enumerate(n):
    if count > 0:
        plt.text(bins[i] + (bins[1] - bins[0])/2, count + 1, str(int(count)),
                 ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

```python
# Основной диапазон значений
max_count_index = n.argmax()
main_range_start = bins[max_count_index]
main_range_end = bins[max_count_index + 1]
{main_range_start} - {main_range_end}
```

```python
# Редкие значения
rare_values = []
for i, count in enumerate(n):
    if count < n.max() * 0.1 and count > 0:
        rare_values.append((bins[i], bins[i+1], count))
for start, end, count in rare_values[:5]:
    print(f"   - Диапазон {start:.0f}-{end:.0f}: {int(count)} треков")
```

```python
# Проверка на пропуски значений
zero_bins = [i for i, count in enumerate(n) if count == 0]
if zero_bins:
    print(f"   - Диапазоны без значений: {', '.join([f'{bins[i]:.0f}-{bins[i+1]:.0f}' for i in zero_bins])}")
else:
    print("   - Нет пустых диапазонов")
```
